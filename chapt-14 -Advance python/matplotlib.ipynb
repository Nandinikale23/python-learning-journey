```python
import matplotlib .pyplot as plt
```

```python
# normal linear graph 
x= [1,2,3,4]
y= [5,6,7,8]
plt.plot(x,y)
plt.show()
```

```python
#for marker to graph and change color 
x=[1,2,3,3]
y=[4,5,8,6]
c= "g"
plt.plot(x,y,c, marker ='.')
plt.show()
```

```python
#show separate x and y axis and with different marker 
x=[1,2,3,3]
y=[4,5,8,6]
#plt.plot(x,y)
plt.plot(x, marker ='.')
plt.plot(y, marker= '>')
plt.show()
```

```python
#Bar graph 
x= [3,2,5,4]
y=[1,4,3,7]
plt.bar(x,y)
plt.show()
```

```python
#change color of bar 
x= [3,2,5,4]
y=[1,4,3,7]
c= ["r", "g", "b","y"]
plt.bar(x,y,color= c)
plt.show()
```

```python
# make bar graph horizontaly
import numpy as np
x= np.array(['a','b','c'])
y= np.array([3,8,5])
c= ["r","b","g"]
plt.barh(x,y,color= c)
plt.show()
```

```python
# set width 
x= [3,2,5,4]
y=[1,4,3,7]

plt.bar(x,y,color= "red", width= 0.2)
plt.show()

```

```python
#change height
x= np.array(['a','b','c'])
y= np.array([3,8,5])
c= ["r","b","g"]
plt.barh(x,y,color= c, height= 0.2)
plt.show()
```

```python
#scatter plot 
x= [1,2,3,4,5]
y= [2,3,4,1,6]
plt.title("scatter plot") 

plt.xlabel("month")
plt.ylabel("number")

c= ["red","blue","pink", "yellow", "orange"] # add different color to every plot
plt.scatter(x,y, color= c, s=200) # to change size of plot using s = size
plt.show()
```

```python
#mapping images to different color 
import numpy as np 
import matplotlib.pyplot as plt 
from PIL import Image 

fname = r"img.jpg.jpg"

#opening image using PIL
image = Image.open(fname).convert("L")

#mapping image to color 
plt.imshow(image, cmap = "gray")
plt.show()
```

```python
#mapping images to different color 
import numpy as np 
import matplotlib.pyplot as plt 
from PIL import Image 

fname = r"img.jpg.jpg"

#opening image using PIL
image = Image.open(fname).convert("L")

#mapping image to color 
plt.imshow(image, cmap = "inferno")
plt.show()
```

```python
#convert into orignal image and show title, grid
import numpy as np 
import matplotlib.pyplot as plt 
from PIL import Image 

fname = r"img.jpg.jpg"

#opening image using PIL
image = Image.open(fname).convert("RGB")

#title
plt.title("Nandini")

#grid
plt.grid()

#mapping image to color 
plt.imshow(image)
plt.show()
```

```python
#piechart
import matplotlib.pyplot as plt
x=[10,30, 40,50]
y=["english","hindi", "math","python"]
plt.pie(x, labels= y)
plt.show()
```

```python
#piechart color change and create label

x=[10,30, 40,50]
y=["english","hindi", "math","python"]
c= ["yellow", "black", "purple", "pink"]
plt.pie(x, labels= y, colors=c)
plt.legend() #for create label chart
plt.show()
```
