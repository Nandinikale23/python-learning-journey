```python
#Bar plot 
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
df2= sns.load_dataset("penguins")
df2
```

```python
df2["island"]
```

```python
sns.barplot(x= df2.island, y= df2.bill_length_mm)
plt.show()
```

```python
#do same thing in 2 way 
sns.barplot(x="island" , y= "bill_length_mm" , data = df2)
plt.show()
```

```python
#hue is used for divides data in categories
sns.barplot(x= "island" , y= "bill_length_mm", data = df2, hue= "sex")
plt.show()
```

```python
# manually chnange order
order1= ["Dream", "Torgersen", "Biscoe"]
sns.barplot(x= "island", y= "bill_length_mm", data = df2, hue= "sex", order= order1)
plt.show()
```

```python
# hue_order it is paramater used for change order like male then female , using hue_order chnage female then male 
order1 = ["Dream", "Torgersen", "Biscoe"]
sns.barplot(x= "island", y= "bill_length_mm", data= df2, hue= "sex", 
            order = order1, hue_order = ["Female", "Male"])
plt.show()
```

```python
order1 = ["Dream", "Torgersen", "Biscon"]
sns.barplot(x= "island", y= "bill_length_mm", data = df2, color= "green")
plt.show()
```

```python
sns.barplot(x= "island", y= "bill_length_mm", data= df2, hue= "sex", palette= "Accent")
plt.show()
```

```python
sns.barplot(x= "island", y= "bill_length_mm", data= df2, hue= "sex", palette= "gist_earth")
plt.show()
```

```python
#saturation := make color faint or dark 0.1  to 0.9
sns.barplot(x= "island", y= "bill_length_mm", data= df2, hue= "sex", palette= "Accent", saturation =0.2)
plt.show()
```

```python
#histogram 
sns.displot(df2["flipper_length_mm"])
plt.show()
```
